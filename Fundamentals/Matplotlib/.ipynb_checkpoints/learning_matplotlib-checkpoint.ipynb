# 1. Introduction to Matplotlib

## 1.1 What is Matplotlib?

**Matplotlib** is the most widely used data visualization library in Python. It allows you to transform raw data into meaningful visualizations such as line charts, bar charts, scatter plots, histograms, pie charts, and many more.

Built on top of **NumPy**, Matplotlib is designed to work seamlessly with NumPy arrays and integrates well with libraries like **Pandas**, **SciPy**, and **Scikit-learn**. It is the standard plotting library used in data analysis, scientific computing, machine learning, and research.

In simple terms, Matplotlib helps you **see patterns, trends, relationships, and outliers in your data** that would be difficult to understand by looking at numbers alone.

# 2. Ways to Create Plots in Matplotlib

Matplotlib provides two main interfaces for creating plots:

1. **Pyplot Interface** – Simple and beginner-friendly.
2. **Object-Oriented Interface** – Recommended for most projects.

You may also encounter a third style that mixes both interfaces, but it is generally discouraged.


### (1) Pyplot Interface

The **Pyplot interface** uses functions from the `matplotlib.pyplot` module to create and customize plots.

### Best For

- Learning Matplotlib
- Quick visualizations
- Small scripts
- Exploratory data analysis (EDA)


````python
plt.plot(x, y)
plt.title("Basic Line Plot Pyplot Interface")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
````

### (2) Object-Oriented Interface  Recommended

The Object-Oriented interface creates a **Figure** and one or more **Axes** objects. All plotting is performed through the **Axes** object.

## Why Use It?

- Recommended by the Matplotlib documentation.
- Provides explicit control over figures and axes.
- Easier to manage multiple plots.
- Standard approach used in data science, machine learning, and production code.


````python
# Method 1: Using add_subplot()

# Create a new Figure object
fig = plt.figure()

# Add a single Axes (subplot) to the Figure
ax = fig.add_subplot()

# Plot the data on the Axes
ax.plot(x, y)

# Display the plot
plt.show()
````

````python
# Method 2: Using add_axes()

# Create a new Figure object
fig = plt.figure()

# Add an Axes at a custom position
# Format: [left, bottom, width, height]
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

# Plot the data on the Axes
ax.plot(x, y)

# Display the plot
plt.show()
````

````python
# Method 3: Using subplots() [Recommended]

# Create a Figure and an Axes object in one line
fig, ax = plt.subplots()

# Plot the data on the Axes
ax.plot(x, y)

# Display the plot
plt.show()
````

# 3. Matplotlib Workflow

##  Matplotlib Workflow Example

````python
# Step 1: Import the required libraries
%matplotlib inline
import matplotlib.pyplot as plt

# Step 2: Prepare the data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 50]

# Step 3: Create a Figure and an Axes object
fig, ax = plt.subplots(figsize=(10,10))  # (width,height)

# Step 4: Plot the data
ax.plot(x, y)

# Step 5: Customize the plot
ax.set(title="Basic Line Plot",
       xlabel="X - axis",
       ylabel="Y - axis"
)

# Step 6: Save and Display the plot
fig.savefig("images/matplotlib-workflow-example.png")
plt.show()
````

# 4. Line Plot

A line plot connects data points using straight lines. It is commonly used to show how values change over time or across a continuous sequence, making it easy to identify trends and patterns.

## (A) Basic Line Plot

````python
# Define the x-axis values
a = [1, 2, 3, 4, 5]

# Define the y-axis values
b = [100, 200, 300, 400, 500]

# Create a Figure and an Axes object
fig, ax = plt.subplots()

# Plot the data
ax.plot(a, b)

# Display the plot
plt.show()
````

## (B) Customizing the Line

Matplotlib allows you to customize the appearance of a line by passing additional parameters to the `plot()` method. You can change properties such as the line color, style, width, and markers to make your plots more informative and visually appealing.

````python
# Create a Figure and an Axes object
fig, ax = plt.subplots(figsize=(8, 5))

# Plot the data with customizations
ax.plot(
    a,
    b,
    color="royalblue",      # Line color
    linestyle="--",         # Dashed line
    linewidth=2.5,          # Line thickness
    marker="o",             # Circle markers
    markersize=8,           # Marker size
    markerfacecolor="white",# Marker fill color
    markeredgecolor="red"   # Marker border color
)

# Add a title and axis labels
ax.set(
    title="Customized Line Plot",
    xlabel="X-axis",
    ylabel="Y-axis"
)

# Display the plot
plt.show()
````

Common `linestyle` options:

```text
'-'    solid [default]
'--'   dashed
'-.'   dash-dot
':'    dotted
```

Common `marker` options:

```text
'o'   circle
's'   square
'^'   triangle
'x'   cross
'*'   star

# 5. Scatter Plot

A scatter plot displays individual data points as dots. It is used to visualize the **relationship or correlation between two numerical variables**, making it easier to identify patterns, trends, clusters, or outliers in the data.

## (A) Basic Scatter Plot

````python
# Importing NumPy
import numpy as np
````

````python
# Creating data for Scatter plot
scatplt_data=np.linspace(0,10,100)
scatplt_data[:10]
````

````python
# Using the data to make a Scatter plot
fig , ax = plt.subplots()
ax.scatter(scatplt_data, np.exp(scatplt_data));
````

## (B) Customising Scatter Plots


````python
# Let's create another scatter plot with the same data
# Using the data to make a Scatter plot
fig , ax = plt.subplots()
ax.scatter(
    scatplt_data, 
    np.exp(scatplt_data));
````
