# 1. Introduction to NumPy

## 1.1 What is NumPy?
NumPy (Numerical Python) is the foundational library for numerical computing in Python. It provides a powerful N-dimensional array object (`ndarray`), along with a large collection of mathematical functions to operate on arrays efficiently. Pandas is built on top of NumPy,every DataFrame column is a NumPy array under the hood.

## (D) np.random
`np.random` is NumPy's module for generating random numbers

Common Functions:
- np.random.rand()    --> Random floats in [0, 1)
- np.random.randint() -->	Random integers
- np.random.random()  -->	Random floats in [0, 1)
- np.random.seed()    -->	Set seed for reproducibility

````python
# 1. np.random.rand(d0, d1, ..., dn)

#where d0, d1, ..., dn are the dimensions of the output array

# Generates random floats between 0 and 1
rand_arr = np.random.rand(2, 3) 
rand_arr
````

````python
# 2. np.random.randint(low, high=None, size=None)

# low  = lowest integer (inclusive)
# high = highest integer (exclusive)
# size = shape of output array

# Generates random integers
randint_arr = np.random.randint(1, 10, size=(2, 3))
randint_arr
````

````python
# 3. np.random.random(size=None)

# size = shape of output array

# Similar to rand(), but takes the shape as a tuple
random_arr = np.random.random((2, 3))
random_arr
````

````python
# 4. np.random.seed(seed

#They are pseudo-random numbers (they look random, but they aren't)
#It allows you to generate random numbers that are reproduceable
#seed = integer used to make results reproducible
````

````python
#lets see an example without seed
print(np.random.randint(1, 10))
print(np.random.randint(1, 10))
````

Different outputs every time because NumPy automatically chooses a seed (usually based on system entropy or time)

````python
np.random.seed(42)
print(np.random.randint(1, 10))
````

 A seed is the initial value that determines the sequence of "random" numbers generated by NumPy.
 Run the program today, tomorrow, or on another computer using the same NumPy version, the sequence starts the same way

 Think of the random generator as a giant book containing a fixed sequence:
 
 Seed 42 → 7, 4, 8, 5, 7, 3, 7, 8, ...
 
 Seed 100 → 9, 9, 4, 8, 8, 1, 5, 2, ...

 Seed 123 → 3, 3, 7, 2, 4, 7, 2, 1, ...

 The seed chooses which page of the book you start reading from

# 5. Manipulating & comparing arrays

## (A) Arithmatic
Arithmetic manipulation refers to performing mathematical operations on the elements of a NumPy array.
NumPy applies these operations element-wise, meaning the operation is performed on each element of the array individually

### (a) Element-wise Arithmetic with Scalars

````python
# Let's perform arithmatic operations on this array
prices = np.array([100.0, 250.0, 430.0, 90.0, 175.0])

print(prices + 10)    # add 10 to every element
print(prices * 1.18)  # apply 18% GST
print(prices ** 2)    # square every element
````
